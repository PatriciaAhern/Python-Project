```python
import random

# Possible train arrival states and actions
states = ["Early", "On Time", "Late"]
actions = ["Slow Down", "Maintain Speed", "Speed Up"]

# Store the agent's scores for each state and action
scores = {
    state: {action: 0 for action in actions}
    for state in states
}

def apply_action(state, action):
    """
    Determine the next train state and reward after an action.
    """

    if state == "Early":
        if action == "Slow Down":
            return "On Time", 10
        elif action == "Maintain Speed":
            return "Early", -2
        else:
            return "Early", -5

    if state == "On Time":
        if action == "Maintain Speed":
            return "On Time", 10
        elif action == "Slow Down":
            return "Late", -5
        else:
            return "Early", -5

    if state == "Late":
        if action == "Speed Up":
            return "On Time", 10
        elif action == "Maintain Speed":
            return "Late", -2
        else:
            return "Late", -5

# Train the agent through repeated trial and error
for episode in range(500):
    current_state = random.choice(states)

    # Explore by trying a random action
    action = random.choice(actions)

    next_state, reward = apply_action(current_state, action)

    # Update the score for the selected action
    scores[current_state][action] += reward

print("Training complete!")
```

```python
for state in states:
    print(f"\nState: {state}")

    for action in actions:
        print(f"{action}: {scores[state][action]} points")
```

```python
print("Best actions learned by the agent:\n")

for state in states:
    best_action = max(scores[state], key=scores[state].get)
    print(f"When the train is {state}, the agent should: {best_action}")
```

```python
test_states = ["Early", "On Time", "Late"]

for state in test_states:
    best_action = max(scores[state], key=scores[state].get)
    next_state, reward = apply_action(state, best_action)

    print(f"Starting state: {state}")
    print(f"Selected action: {best_action}")
    print(f"New state: {next_state}")
    print(f"Reward: {reward}")
    print("-" * 30)
```
